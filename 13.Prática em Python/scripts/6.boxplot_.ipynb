```python
# Importação das bibliotecas
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Carregamento da base de dados
base = pd.read_csv('../dados/trees.csv')
base
```

```python
# Geração do boxplot
# patch_artist = True preenche, show fliers outliers
plt.boxplot(base.Volume, vert=False, showfliers=False, notch=True, patch_artist=True)
plt.title('Árvores')
plt.xlabel('Volume')
plt.show()
```

```python
# Dados por linha
plt.boxplot(base)
plt.title('Árvores')
plt.xticks(ticks=[1, 2, 3], labels=[*base.columns.tolist()])
plt.show()
```

```python
# Geração de 3 boxplot, cada um mostrado informações diferentes
plt.figure(figsize=[15, 10])
for i, column in enumerate(base.columns, start=1):
    plt.subplot(2, 2, i)
    plt.boxplot(base[column], vert=False, notch=True, patch_artist=True)
    plt.title(column)
    plt.yticks([])
plt.subplots_adjust(wspace=0.3, hspace=0.3)
plt.show()
```
