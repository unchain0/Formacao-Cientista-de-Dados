```python
# Importação do pandas para leitura de arquivos .csv
import pandas as pd
```

```python
# Carregamento da base de dados
base = pd.read_csv('../dados/insect.csv')
base.shape
```

```python
# Dados
base
```

```python
# Agrupamento dos dados baseado no atributo 'spray', contando e somando os registros
agrupado: pd.Series = base.groupby(['spray'])['count'].sum()
agrupado.sort_values(ascending=False)
```

```python
# Gráfico de barras
agrupado.plot.bar(color='gray', rot=0)
```

```python
# Cores
agrupado.plot.bar(color=['blue', 'yellow', 'red', 'green', 'pink', 'orange'], rot=0)
```

```python
# Gráfico de pizza
agrupado.plot.pie()
```

```python
# Com legenda
agrupado.plot.pie(legend=True)
```
