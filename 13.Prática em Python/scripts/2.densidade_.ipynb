```python
# Importação das bibliotecas
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
# Carregamento da base de dados
base = pd.read_csv('../dados/trees.csv')
base.head()
```

```python
girth_height = base.iloc[:, 1]
```

```python
# Histograma com 10 divisões (bins) e somente para o primeiro atributo da base de dados
plt.hist(girth_height, bins=6)
```

```python
# Histograma com a linha de distribuição de frequência, com 6 divisões (bins)
# kde = linha de densidade
sns.histplot(girth_height, bins=6, color='blue').set(title="Árvores")
```

```python
# Densidade
sns.kdeplot(girth_height, color='blue').set(title='Árvores')
```

```python
# Densidade e histograma
sns.histplot(girth_height, kde=True, bins=6, color='blue').set(title='Árvores')
```
