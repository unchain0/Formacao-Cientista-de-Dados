```python
# Importação das bibliotecas
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
# Carregamento da base de dados
base = pd.read_csv('../dados/trees.csv')
base
```

```python
# Visualização de um boxplot
plt.figure(figsize=(3, 5))
sns.boxplot(data=base.Volume, width=0.4, color='blue')
plt.title('Árvores')
plt.xticks(ticks=[0], labels=['Volume'])
plt.show()
```

```python
# Calculando IQR das colunas da base
for column in base.columns:
    q1 = base[column].quantile(0.25)
    q3 = base[column].quantile(0.75)
    iqr = q3 - q1
    iqr = round(iqr, 2)
    print(f"{column=:<10}{iqr=}")
```

```python
# Visualização de vários boxplot na mesma imagem
plt.figure(figsize=(5, 7))
sns.boxplot(data=base, palette='coolwarm', width=0.6)
plt.title('Árvores')
plt.show()
```
