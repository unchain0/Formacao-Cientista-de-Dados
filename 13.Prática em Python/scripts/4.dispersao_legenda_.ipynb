```python
# Importação das bibliotecas
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Carregamento da base de dados
base = pd.read_csv('../dados/co2.csv')
base
```

```python
# Criando duas variáveis para cada atributo (x = concentration e y = uptake)
x = base.conc
y = base.uptake
```

```python
# Retorna os valores únicos do atributo "treatment"
unicos = list(set(base.Treatment))
unicos
```

```python
# Percorre cada tipo de tratamento (chilled e nonchilled) e cria um gráfico de dispersão
for i in range(len(unicos)):
    index = base.Treatment == unicos[i]
    plt.scatter(x[index], y[index], label=unicos[i])
plt.xlabel("Concentration")
plt.ylabel("Treatment")
plt.legend(loc='lower right')
plt.show()
```
