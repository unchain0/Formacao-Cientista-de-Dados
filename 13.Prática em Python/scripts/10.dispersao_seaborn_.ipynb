```python
# Importação das bibliotecas
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import warnings
```

```python
warnings.simplefilter(action='ignore', category=UserWarning)
```

```python
# Carregamento da base de dados
base = pd.read_csv('../dados/co2.csv')
base
```

```python
# Gráfico de dispersão utilizando os atributos conc e uptake, agrupamento pelo type
sns.scatterplot(x=base.conc, y=base.uptake, hue=base.Type, palette='hsv')
plt.show()
```

```python
# Seleção de registros específicos da base de dados (Quebec e Mississippi)
q = base.loc[base['Type'] == 'Quebec']
m = base.loc[base['Type'] == 'Mississippi']
```

```python
# Sub gráfico (1 linha e duas colunas) mostrando gráficos sobre cada região
plt.figure(figsize=[10, 5])
plt.subplot(1, 2, 1)
sns.scatterplot(x=q.conc, y=q.uptake).set_title('Quebec')
plt.subplot(1, 2, 2)
sns.scatterplot(x=m.conc, y=m.uptake).set_title('Mississippi')
plt.tight_layout(pad=2.5)
```

```python
# Refrigerado e não refrigerado
ch = base.loc[base['Treatment'] == 'chilled']
nc = base.loc[base['Treatment'] == 'nonchilled']
```

```python
# Gráfico somente com 'chilled' e 'nonchilled'
plt.figure(figsize=[10, 5])
plt.subplot(1, 2, 1)
sns.scatterplot(x=ch.conc, y=ch.uptake).set_title('Chilled')
plt.subplot(1, 2, 2)
sns.scatterplot(x=nc.conc, y=nc.uptake).set_title('Non chilled')
plt.tight_layout(pad=2.5)
```

```python
# Carregamento de outro arquivo, câncer de esôfago
base2 = pd.read_csv('../dados/esoph.csv')
base2
```

```python
# Gráfico entre os atributos 'alcgp' e 'ncontrols'
sns.catplot(x='alcgp', y='ncontrols', data=base2, jitter=False, hue='ncontrols', palette='hsv')
plt.title("Esôfago")
plt.show()
```

```python
# Gráfico entre os atributos 'alcgp' e 'ncontrols', com agrupamento
sns.catplot(x='alcgp', y='ncontrols', data=base2, col='tobgp', hue='ncontrols', palette='hsv', aspect=0.62)
plt.suptitle('Relação entre alcgp e ncontrols por tobgp', fontsize=16)
plt.tight_layout(w_pad=1.5)
plt.show()
```
