```python
# Importação das bibliotecas
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
# Carregamento da base de dados
base = pd.read_csv(r'..\dados\trees.csv')
base.head()
```

```python
# Histograma com 10 divisões (bins) e com gráfico de densidade
sns.histplot(data=base.Volume, bins=10, kde=True)
plt.title("Árvores")
plt.show()
```

```python
# Carregamento de outra base de dados
base2 = pd.read_csv(r'..\dados\chicken.csv')
base2
```

```python
# Criação de novo dataframe agrupando o atributo 'feed'
agrupado = base2.groupby(['feed'])['weight'].sum()
agrupado
```

```python
# Novo dataframe somente para testar os filtros do pandas
teste = base2.loc[base2['feed'] == 'horsebean']
teste
```

```python
# Histograma considerando todos os valores
unique_values = base2['feed'].unique().tolist()
for value in unique_values:
    sns.histplot(base2.loc[base2['feed'] == value].weight, kde=True)
    plt.title(value.title())
    plt.show()
```

```python
# Aumenta o tamanho das figuras
plt.figure(figsize=[10, 10])
# Impressão em gráfico 2 x 3
for i, value in enumerate(unique_values, start=1):
    plt.subplot(3, 2, i)
    sns.histplot(base2.loc[base2['feed'] == value].weight, kde=True, color=sns.color_palette()[i])
    plt.title(value.title())
# Ajusta o layout para não haver sobreposição
plt.tight_layout(pad=2.5)
```
