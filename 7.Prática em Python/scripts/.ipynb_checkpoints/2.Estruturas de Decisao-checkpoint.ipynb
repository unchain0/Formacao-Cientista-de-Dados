Estruturas de Decisão

```python
try:
    nota = float(input('Nota do aluno: '))
    freq = float(input('Frequência (%): '))
    if 0 <= nota <= 10 and 0 <= freq <= 100:
        if nota >= 5 and freq >= 70:
            print('Aprovado!')
        else:
            print('Reprovado!')
    else:
        raise ValueError
except ValueError:
    print('Nota ou frequência inválida(s).')
```

```python
try:
    nota = float(input('Nota do aluno: '))
    if 0 <= nota <= 10:
        if nota <= 4:
            print('Reprovado!')
        elif nota <= 6:
            print('Recuperação!')
        else:
            print('Aprovado!')
    else:
        raise ValueError
except ValueError:
    print('Nota inválida.')
```
