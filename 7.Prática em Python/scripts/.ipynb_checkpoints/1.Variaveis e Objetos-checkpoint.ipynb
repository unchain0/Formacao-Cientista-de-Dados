Variáveis e Objetos

```python
# Cria variável do tipo inteiro
x = 1
print(x, type(x))
# Cria variável do tipo float
y = 3.13
print(y, type(y))
# Cria variável do tipo string
z = "Python"
print(z, type(z))
z = 'Python'
print(z, type(z))
# Cria variável do tipo lógico
w = True
y = False
print(w, type(w))
print(y, type(y))
```

```python
# Calculadora
x = 10
y = 20
z = 100
w = (x + y) * z / 100
print(f'O total é {w:.0f}')
```

```python
# Exibir valores no console
print('Este texto será impresso no console')
print(x)
print(f'Texto e duas variáveis {x} {z}')
```

```python
# Verificar tipo
x = 10
print(x, type(x))
y = 'Python'
print(y, type(y))
```

```python
# Entrada de valores
x = input('Informe um valor: ')
print(x, type(x))
```

```python
# Conversão de Valores
x = int(z)
w = str(y)
t = float(x)
```
