```python
# Com retorno
def potencia(n):
    return n * n


print(potencia(3))
```

```python
# Com valor default
def intervalo(ini=1, fim=10):
    for n in range(ini, fim + 1):
        print(n)


x = intervalo(fim=3)
y = intervalo()
```
