Estruturas de Repetição

```python
# Exemplo com while
count = 1
while count <= 5:
    print(count, end=', ')
    count += 1
print('Fim')
```

```python
# range(0, 10, 1):
# range(inicio, parada, incremento)
# O Python "ignora" o último valor de intervalos
# O incremento pode ser negativo

# for normal
for n in range(0, 10):
    print(n + 1, end=', ')
print('Fim')
for n in range(1, 11):
    print(n, end=', ')
print('Fim')
```

```python
# for decrescente
for n in range(10, 0, -1):
    print(n, end=', ')
print('Fim')
```

```python
# Exemplo com break
# Vai parar no número digitado pelo usuário
x = int(input('Digite um número de 0-10:'))
for n in range(11):
    if n == x:
        print(f'Parar no {n}', end=', ')
        break
    print(n, end=', ')
print('Fim')
```

```python
# Exemplo com continue
# Vai pular o número digitado pelo usuário
x = int(input('Digite um número de 0-10:'))
for n in range(11):
    if n == x:
        print(' ', end=', ')
        continue
    print(n, end=', ')
print('Fim')
```
