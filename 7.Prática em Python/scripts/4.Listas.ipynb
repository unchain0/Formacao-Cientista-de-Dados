Listas

```python
# Formas de criar listas
lst = list(range(1, 6))
print(lst)
```

```python
# Vários tipos
lst2 = [1, 2, 3, "4", True]
print(lst2)
```

```python
# Lista com lista
lst3 = [12, list(range(1, 6)), "a"]
print(lst3)
```

```python
# Criando lista com range
lst4 = list(range(10))
print(lst4)
```

```python
# Comprimento da lista
print(lst, f"Tamanho: {len(lst)}")
```

```python
# Acessando primeiro elemento
print(lst[0])
# Acessando último elemento
print(lst[-1])
# Índice 1 é uma lista
print(lst3[1])
# Alterando valor pelo índice
lst[0] = 5
print(lst)
```

```python
# Percorrendo uma lista
print(f"Lista 4 contém {len(lst4)} elementos, são eles: ", end="")
for n in lst4:
    print(n + 1, end=', ')
print('FIM!')
```
