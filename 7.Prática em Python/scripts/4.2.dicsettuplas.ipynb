```python
# Dicionários: chave: valor
Notas = {'João': 6.0, 'Maria': 8.0, 'Pedro': 6.5}
print(Notas, len(Notas))
```

```python
print(Notas['João'])
```

```python
# Todas as chaves
print(Notas.keys())
```

```python
# Todos os valores
print(Notas.values())
```

```python
# Testa logicamente a existência
print('João' in Notas)
print('Fernando' in Notas)
```

```python
# Apaga um elemento
del Notas['João']
print(Notas)
```

```python
# Inclui um elemento
Notas['Ana'] = 9
print(Notas)
```

```python
# Busca um valor, se não encontrar mostra o segundo parâmetro
print(Notas.get('Geraldo', 'Não encontrado!'))
print(Notas.get('Ana', 'Não encontrado!'))
```

```python
# ‘sets’ são conjuntos não ordenados de elementos não repetidos
bigdata = {'Spark', 'Hive', 'Sqoop'}
print(bigdata)
```

```python
# Verifica existência
print('Spark' in bigdata)
print('Python' in bigdata)
```

```python
# Adiciona elemento
bigdata.add('Hadoop')
bigdata
```

```python
# Número de elementos
print(len(bigdata))
```

```python
bigdata.add("Spark")
print(bigdata)
```

```python
# Tuplas são listas que não podem ser alteradas
# Listas usam [], tuplas usam ()
tupla = tuple(range(1, 8))
print(tupla)
```

```python
print(tupla[4])
```

```python
# Dicionários em que cada posição recebe uma tupla
dic2 = {(0, 1): 0, (1, 2): 1, (2, 3): 2, (3, 4): 3, (4, 5): 4,(5, 6): 5, (6, 7): 6, (7, 8): 7, (8, 9): 8, (9, 10): 9}
dic2
```

```python
print(type(Notas))
print(type(bigdata))
print(type(tupla))
```

```python
# Percorrendo dicionários (dict)
for k, v in Notas.items():
    print(f'\033[33m{k}\033[m | \033[32m{v}\033[m')
```

```python
# Percorrendo conjuntos (set)
for v in bigdata:
    print(v)
```

```python
# Percorrendo tuplas (tuple)
print(tupla, end=f'\n{"-=" * 11}\n')
for v in tupla:
    print(f'Index: {tupla.index(v)} | Value: {v}')
print('-=' * 11)
for i, v in enumerate(tupla):
    print(f'Index: {i} | Value: {v}')
```
