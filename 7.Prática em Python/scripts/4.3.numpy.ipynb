```python
import numpy as np
```

```python
# Cria uma matriz unidimensional
mt = np.array([12, 34, 26, 18, 10])
print(mt, type(mt))
```

```python
# Cria o array com um tipo específico
# Cria o array como float de 64 bits
mtfloat = np.array([1, 2, 3], dtype = np.float64)
print(mtfloat, type(mtfloat))
mtint = np.array([1, 2, 3], dtype = np.int32)
print(mtint, type(mtint))
```

```python
# Mudar o tipo do array
# Podemos transformar tipos de dados de arrays
mtnew = np.array([1.4, 3.6, -5.1, 9.42, 4.999999])
print(mtnew)
# Quando transformamos de float para inteiro os valores são truncados
mtnewint = mtnew.astype(np.int32)
print(mtnewint)
```

```python
# Podemos fazer o inverso também.
mt5 = np.array([1, 2, 3, 4])
print(mt5)
mt6 = mt5.astype(float)
print(mt6)
```

```python
# Mais de uma dimensão
# Cria um matriz bidimensional
mt7 = np.array([[7, 2, 23], [12, 27, 4], [5, 34, 23]])
print(mt7)
```

```python
# Criar arrays vazios tipificados
# Empty significa que não são inicializados, não que são vazios
vazio = np.empty([3, 2], dtype=int)  # 3 linhas, 2 colunas
print(vazio)
print('-=' * 15)
# Cria uma matriz 4x3 com valores zero
zeros = np.zeros([4, 3], dtype=int)
print(zeros)
print('-=' * 15)
# Com valor igual a um
um = np.ones([5, 7], dtype=int)
print(um)
print('-=' * 15)
# Cria matriz quadrada com diagonal principal com valores 1 e os outros valores zero
diagonal = np.eye(5)  # 5x5
print(diagonal)
```

```python
# Valores aleatórios entre zero e um
ale = np.random.random(5)
print(ale)
print('-=' * 30)
# Valores aleatórios distribuição normal contendo negativos
ale2= np.random.randn(5)
print(ale2)
print('-=' * 30)
# Valores positivos 1+ aleatórios 3x4
ale3 = (10 ** np.random.random([3, 4]))
print(ale3)
```

```python
# Outra forma de gerar aleatórios
# Uso de semente
gnr = np.random.default_rng(1)
print(gnr, type(gnr))
print('-=' * 30)
ale5 = gnr.random(3)
print (ale5)
print('-=' * 30)
# Gerar inteiros
ale6 = gnr.integers(2, size=[5, 15])
print(ale6)
print('-=' * 30)
ale7 = gnr.integers(100, size=(5, 15))
print(ale7)
print('-=' * 30)
ale7 = np.unique(ale7)
print(ale7)
```

```python
# Unique remove repetições
j = np.array([14, 24, 11, 12, 13, 14, 15, 16, 21, 17, 12, 13, 11, 18, 19, 20])
print(j)
j = np.unique(j)  # Também ordena os valores em ordem crescente
print(j)
```

```python
# Função específica
# Cria a matriz bidimensional k
k = np.array([[17, 22, 43], [27, 25, 14], [15, 24, 32]])
# Mostra a matriz k
print(k)
# Mostra um elemento específico da matriz k
print(k[0][1])  # 0 - Linha | 1 - Coluna
print(k.shape)  # Formato da matriz k
```

```python
# Lendo as linhas da matriz k
for l in k:  # Linhas
    print(l)
print('-=' * 15)
# Lendo os valores da matriz k
for l in k:  # Linhas
    for v in l:  # Colunas
        print(f'[{v:^6}]', end='')
    print()
```

```python
# Funções Matemáticas
# Mostra o maior valor da matriz k
print(k.max())
# Mostra o menor valor da matriz k
print(k.min())
# Mostra a soma dos valores da matriz k
print((k.sum()))
# Mostra o valor da média dos valores da matriz k
print(k.mean())
# Mostra o valor do desvio padrão (‘standard’ deviation) dos valores da matriz k
print(k.std())
```

```python
# Funções universais, aplicadas a todos os elementos
# Mostra o valor da raiz quadrada de todos elementos
k1 = np.array([1, 4, 9, 16, 25, 36])
print(np.sqrt(k1))
# Mostra o valor do exponencial de todos elementos
print(np.exp(k1))
```

```python
# Extração de elementos
m = gnr.integers(10, size=6)
print(m)
print('-=' * 7)
# Mostra o elemento da posição 2
print('[1]', m[1])
print('-=' * 7)
# Mostra o array criado a partir da posição 0, dois elementos
print('[:2]', m[:2])
print('-=' * 7)
# Mostra o array criado a partir do 1.º índice até o final do array
print('[1:]', m[1:])
print('-=' * 7)
# Mostra o array criado a partir da antepenúltima posição até o final
print('[-3:]', m[-3:])
```

```python
# Extração de linhas e colunas
l = gnr.integers(10, size=(3, 2))
print(l)
print("-=" * 17)
print('Primeira linha, todas as colunas')
l_linha_1 = l[0, :]
print(l_linha_1)
print("-=" * 17)
print('Segunda linha, todas as colunas')
l_linha_2 = l[1, :]
print(l_linha_2)
print("-=" * 17)
print('Terceira linha, todas as colunas')
l_linha_3 = l[2, :]
print(l_linha_3)
print("-=" * 17)
print('Todas as linhas, primeira coluna')
l_coluna_1 = l[:, 0]
print(l_coluna_1)
print("-=" * 17)
print('Todas as linhas, segunda coluna')
l_coluna_2 = l[:, 1]
print(l_coluna_2)
```

```python
# Adição e multiplicação de matrizes
n = np.array([[1, 2], [3, 4]])
o = np.array([[1, 1], [1, 1]])
res1 = n+o
print(res1)
print("--------")
res2 = n*o
print(res2)
print("--------")
p = np.array([[1, 2], [3, 4], [5, 6]])
q = np.array([2, 1])
print(p+q)
```

```python
# Transposição, rearranja um conjunto de 15 elementos de 0 a 14 em 3 linhas e 5 colunas.
f = np.arange(15).reshape((3, 5))
# Mostra a matriz transposta entre linha e coluna
print(f)
print("-------------------")
s = f.T
print(s)
```

```python
# Outra forma de fazer, mesmo resultado
r = np.arange(15).reshape((3, 5))
print(r)
print("-------")
# Rearranja um conjunto de 15 elementos, mostra a matriz transposta entre linha e coluna
s = r.transpose((1, 0))
print(s)
```

```python
# Expressões lógicas
# Usando where
# Criando matriz com valores aleatórios positivos e negativos
v = np.random.randn(4, 4)
print(v)
# Criando matriz com valores booleanos baseado no array v
x = v > 0
print(x)
# Criando matriz com valores 0 e 1 baseado nos valores do array x
z = np.where(x > 0, 1, 0)
print(z)
```
